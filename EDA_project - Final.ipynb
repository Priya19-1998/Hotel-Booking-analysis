# **VISUALIZATION**

**OVERALL CANCELLATION RATE**

Two bars:

0 — Not Cancelled: around 75,000 bookings
1 — Cancelled: around 44,000 bookings

```python
df['is_canceled'].value_counts().plot(
    kind='bar',
    title='Overall Booking Cancellation Status'
)

plt.xlabel('Booking Status (0 = Not Cancelled, 1 = Cancelled)')
plt.ylabel('Number of Bookings')
plt.show()
```

**City Hotel vs Resort Hotel cancellation**

City Hotel → 41.73%
Resort Hotel → 27.76%

```python
df.groupby('hotel')['is_canceled'].mean().mul(100).plot(
    kind='bar',
    title='Cancellation Rate by Hotel Type'
)

plt.xlabel('Hotel Type')
plt.ylabel('Cancellation Rate (%)')
plt.xticks(rotation=0)
plt.show()
```

**Lead Time vs Cancellation**

Not cancelled: ~79.98 days

Cancelled: ~144.85 days

```python
df.groupby('is_canceled')['lead_time'].mean().plot(
    kind='bar',
    title='Average Lead Time by Booking Status'
)

plt.xlabel('Booking Status (0 = Not Cancelled, 1 = Cancelled)')
plt.ylabel('Average Lead Time (Days)')
plt.xticks(rotation=0)
plt.show()
```

**Previous Cancellation History**

No → 33.91%

Yes → 91.64%

```python
df.groupby('has_previous_cancellation')['is_canceled'].mean().mul(100).plot(
    kind='bar',
    title='Cancellation Rate by Previous Cancellation History'
)

plt.xlabel('Previous Cancellation History')
plt.ylabel('Cancellation Rate (%)')
plt.xticks(rotation=0)
plt.show()
```

**Previous Successful Booking vs Cancellation**

No previous successful booking: 38.03% cancellation

Has previous successful booking: 5.52% cancellation


```python
df.groupby('has_previous_booking')['is_canceled'].mean().mul(100).plot(
    kind='bar',
    title='Cancellation Rate by Previous Booking History'
)

plt.xlabel('Previous Successful Booking')
plt.ylabel('Cancellation Rate (%)')
plt.xticks(rotation=0)
plt.show()
```

**Waiting List vs Cancellation**

No waiting-list days: 36.19% cancellation

Had waiting-list days: 63.79% cancellation

```python
df.groupby('has_waiting_list')['is_canceled'].mean().mul(100).plot(
    kind='bar',
    title='Cancellation Rate by Waiting List Status'
)

plt.xlabel('Waiting List Status')
plt.ylabel('Cancellation Rate (%)')
plt.xticks(rotation=0)
plt.show()
```

**Distribution Channel**

# **SUBJECTIVE QUESTIONS**

```python
df['lead_time_group'] = pd.cut(
    df['lead_time'],
    bins=[-1, 30, 60, 90, 180, float('inf')],
    labels=['0-30 days', '31-60 days', '61-90 days', '91-180 days', '181+ days']
)
```

```python
df.groupby('lead_time_group', observed=True)['is_canceled'].mean() * 100
```

```python
df.groupby('lead_time_group', observed=True)['is_canceled'].mean().mul(100).plot(
    kind='bar',
    title='Cancellation Rate by Lead Time Group'
)

plt.xlabel('Lead Time')
plt.ylabel('Cancellation Rate (%)')
plt.xticks(rotation=0)
plt.show()
```
